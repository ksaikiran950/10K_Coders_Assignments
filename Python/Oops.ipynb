```python
class Calci:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def sub(self):
        return self.a - self.b

    def mul(self):
        return self.a * self.b

    def div(self):
        return self.a / self.b

    def rem(self):
        return self.a % self.b


# create object
s1 = Calci(100, 20)

print("Addition:", s1.add())
print("Subtraction:", s1.sub())
print("Multiplication:", s1.mul())
print("Division:", s1.div())
print("Remainder:", s1.rem())
```

```python
class Calci:

    def add(self,a,b):
        return a + b

    def sub(self,a,b):
        return a - b

    def mul(self,a,b):
        return a * b

    def div(self,a,b):
        return a / b

    def rem(self,a,b):
        return a % b


print("Choose operation: ")
print("1. Add\n2. Sub\n3. Mul\n4. Div\n5. Rem\n6. Exit")

s1 = Calci()

while True:
    n = int(input("\nEnter your option: "))

    if n == 6:
        print("Exit")
        break

    a, b = map(int, input("Enter two values : ").split())

    if n == 1:
        print("Result:", s1.add(a,b))
    elif n == 2:
        print("Result:", s1.sub(a,b))
    elif n == 3:
        print("Result:", s1.mul(a,b))
    elif n == 4:
        print("Result:", s1.div(a,b))
    elif n == 5:
        print("Result:", s1.rem(a,b))
    else:
        print("Invalid choice. Please try again.")
```

```python
class Calci:
    def __init__(self, name):
        self.name = name  

    def get_name(self):
        return self.name

    def add(self, a, b):
        return a + b

    def sub(self, a, b):
        return a - b

    def mul(self, a, b):
        return a * b

    def div(self, a, b):
        return a / b

    def rem(self, a, b):
        return a % b


# Step 1: Take user name
name = input("Enter your name: ")

# Step 2: Create object
s1 = Calci(name)

# Step 3: Greet after object is created
print(f"\nHello {s1.get_name()}! Welcome ")

print("Choose operation: ")
print("1. Add\n2. Sub\n3. Mul\n4. Div\n5. Rem\n6. Exit\n7. Show Name")

while True:
    n = int(input("\nEnter your option: "))

    if n == 6:
        print(f"Exit")
        break

    elif n == 7:
        print(f"Name: {s1.get_name()}")
        continue

    a, b = map(int, input("Enter two values : ").split())

    if n == 1:
        print("Result:", s1.add(a, b))
    elif n == 2:
        print("Result:", s1.sub(a, b))
    elif n == 3:
        print("Result:", s1.mul(a, b))
    elif n == 4:
        print("Result:", s1.div(a, b))
    elif n == 5:
        print("Result:", s1.rem(a, b))
    else:
        print("Invalid choice. Please try again.")
```
