```python
class greeting:
    def __init__(self,name):
        self.name=name
    def visit(self,name):
        print(f"hello {name}   sir!  ")
    def exit(self):
        print(f"{self.name} TQ visit again !")

g1=greeting("sai")
g1.visit("kiran")
g1.exit()
```

```python
class calci:
    def __init__(self,first,second):
        self.first=first
        self.second=second
    def add(self):
        return self.first+self.second
    def sub(self,a,b):
        return a-b
    
c1=calci(10,20)
c2=calci(100,88)
print(c1.add())
print(c2.sub(1,2))
```
