```python
def prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

def add(n):
    if n==0:
        return n
    return (n%10)+add(n//10)

x=10
y=20
for i in range(x,y+1):
    num=add(i)
    if prime(num):
        print(i)
```

```python
def fib(n):
    if n==0 or n==1:
        return n
    return fib(n-1)+fib(n-2)

n=4
for i in range(n):
    for j in range(i+1):
        print(fib(i),end=" ")
    print()
```

```python
n = 4

a, b = 0, 1  # first two Fibonacci numbers

for i in range(1, n + 1):
    # reset starting points differently for each row
    x, y = a, b
    for j in range(1, i + 1):
        print(x, end=" ")
        x, y = y, x + y  # next Fibonacci number
        a, b = b, a + b  # global Fibonacci progress
    print()
```

```python
def print_fib_pattern():
    n = 10
    fib = [0, 1]
    
    for i in range(2, n):
        next_fib = fib[i-1] + fib[i-2]
        fib.append(next_fib)
        
    num_rows = 4
    
    for i in range(num_rows):
        
        current_row_numbers = []
        
        for j in range(i + 1):
            
            if j == 0:
                current_row_numbers.append(fib[i])
            elif j == 1:
                current_row_numbers.append(fib[i + 3])
            elif j == 2:
                current_row_numbers.append(fib[i + 5])
            elif j == 3:
                current_row_numbers.append(fib[i + 6])
                
        print(" ".join(map(str, current_row_numbers)))

print_fib_pattern()
```
