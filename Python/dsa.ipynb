```python
#s="a[1]b[2]c[3]"
#s="ab[3]b[2]c[1]"
s="ab[10]b[1]"
res=""
alpha=""
digit=""

for i in range(len(s)):
    if s[i] =="]":
        res+=alpha*int(digit)
        alpha=""
        digit=""
    elif(s[i].isdigit()):
        digit+=s[i]
    elif(s[i].isalpha()):
        alpha+=s[i]

print(res)


```
