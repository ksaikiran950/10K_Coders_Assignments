```python
s="music reduces depression"
def count(s):
    d={"v":0,"c":0}
    for ch in s:
        if ch.isalpha():
            if ch in "aeiouAEIOU":
                d["v"]+=1
            else:
                d["c"]+=1
    return d


dict1=count(s)
print(f'vowels: {dict1["v"]}, consonants: {dict1["c"]}')
```

```python
def slargest(l1):
    fmax=smax=float("-inf")
    for i in l1:
        if i>fmax:
            smax=fmax
            fmax=i
        elif i>smax and i !=fmax:
            smax=i

    return smax

print("second largest :",slargest(l1))




l1=[2,6,7,67,87,9]
```
